```python
sns.countplot(data=df,x='Ticket Type',order=df['Ticket Type'].value_counts().index)
plt.xticks(rotation=45)
plt.show()
```

```python
sns.countplot(data=df,x='Ticket Priority',order=df['Ticket Priority'].value_counts().index)
plt.show()
```

```python
sns.countplot(data=df,x='Ticket Channel',order=df['Ticket Channel'].value_counts().index)
plt.show()
```

```python
sns.histplot(df['Customer Satisfaction Rating'],bins=5,kde=True)
plt.show()
```

```python
sns.histplot(df['Time to Resolution'],bins=30,kde=True)
plt.show()
```

```python
df['Description_Length']=df['Ticket Description'].astype(str).apply(len)
sns.histplot(df['Description_Length'],bins=30,kde=True)
plt.show()
```

```python
df['Subject_Length']=df['Ticket Subject'].astype(str).apply(len)
sns.histplot(df['Subject_Length'],bins=30,kde=True)
plt.show()
```

```python
text=' '.join(df['Ticket Description'].dropna().astype(str))
wc=WordCloud(width=1000,height=500,background_color='white').generate(text)
plt.imshow(wc)
plt.axis('off')
plt.show()
```

```python
text=' '.join(df['Ticket Subject'].dropna().astype(str))
wc=WordCloud(width=1000,height=500,background_color='white').generate(text)
plt.imshow(wc)
plt.axis('off')
plt.show()
```
