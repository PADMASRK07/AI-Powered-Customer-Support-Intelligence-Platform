```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

df = pd.read_csv('D:\Customer Support\Data set\customer_support_tickets_clean.csv')
df_fe = df.copy()
```

```python
cat_cols = ['Ticket Channel','Product Purchased','Customer Gender']
encoders={}
for col in cat_cols:
    le=LabelEncoder()
    df_fe[col]=le.fit_transform(df_fe[col].astype(str))
    encoders[col]=le

df_fe[cat_cols].head()
```

```python
df_fe['Date of Purchase']=pd.to_datetime(df_fe['Date of Purchase'],errors='coerce')
df_fe['Purchase_Year'] = df_fe['Date of Purchase'].dt.year
df_fe['Purchase_Month'] = df_fe['Date of Purchase'].dt.month
df_fe['Purchase_Day'] = df_fe['Date of Purchase'].dt.day
df_fe['Purchase_Weekday'] = df_fe['Date of Purchase'].dt.dayofweek
```

```python
df_fe['Description_Char_Count']=df_fe['Ticket Description'].fillna('').astype(str).apply(len)
df_fe['Description_Word_Count']=df_fe['Ticket Description'].fillna('').astype(str).apply(lambda x: len(x.split()))
```

```python
df_fe['Subject_Char_Count']=df_fe['Ticket Subject'].fillna('').astype(str).apply(len)
df_fe['Subject_Word_Count']=df_fe['Ticket Subject'].fillna('').astype(str).apply(lambda x: len(x.split()))
```

```python
df_fe.to_csv('D:\Customer Support\Data set\customer_support_tickets_FE.csv',index=False)
print(df_fe.head())
```
